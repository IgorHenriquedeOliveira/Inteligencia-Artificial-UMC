```python
# Exercício 22
def cifra_cesar(mensagem, deslocamento):
    resultado = ""
    for letra in mensagem:
        if letra.isalpha():
            codigo = ord(letra) + deslocamento
            resultado = resultado + chr(codigo)
        else:
            resultado = resultado + letra
    return resultado
texto = input("Digite a mensagem: ")
print("Criptografado:", cifra_cesar(texto, 3))
# usei ord() pra virar numero da tabela ascii, somei o deslocamento e chr() pra virar letra de novo, e o isalpha pra ignorar espaço ou pontuacao.
```

```python
# Exercício 23
def descifra_cesar(mensagem_criptografada, deslocamento):
    resultado = ""
    for letra in mensagem_criptografada:
        if letra.isalpha():
            codigo = ord(letra) - deslocamento
            resultado = resultado + chr(codigo)
        else:
            resultado = resultado + letra
    return resultado
print("Descriptografado:", descifra_cesar("def", 3))
# é o mesmo codigo da criptografia, a diferença é que eu diminui o deslocamento no ord() em vez de somar pra voltar ao que era.
```

```python
# Exercício 24
def valida_cpf(cpf):
    cpf = cpf.replace(".", "").replace("-", "")
    if len(cpf) != 11:
        return False
    soma = 0
    peso = 10
    for i in range(9):
        soma = soma + int(cpf[i]) * peso
        peso = peso - 1
    resto = (soma * 10) % 11
    if resto == 10:
        resto = 0
    if resto != int(cpf[9]):
        return False
    return True
meu_cpf = input("Digite o CPF: ")
if valida_cpf(meu_cpf):
    print("CPF válido!")
else:
    print("CPF inválido.")
# tirei o ponto e traço com replace pra limpar, e fiz um for pra ir multiplicando pelos pesos e achar o primeiro digito verificador pelo menos.
```

```python
# Exercício 31
lista = [5, 1, 4, 2, 8]
tamanho = len(lista)
for i in range(tamanho):
    for j in range(0, tamanho - 1):
        if lista[j] > lista[j+1]:
            temp = lista[j]
            lista[j] = lista[j+1]
            lista[j+1] = temp
print("Lista ordenada:", lista)
# bubble sort feito na mao com dois for aninhados, se o atual for maior que o proximo eu uso uma variavel temp pra trocar os dois de lugar na lista.
```

```python
# Exercício 46
import math
lista = [10, 12, 23, 23, 16, 23, 21, 16]
soma = 0
for n in lista:
    soma = soma + n
media = soma / len(lista)
soma_diferencas = 0
for n in lista:
    soma_diferencas = soma_diferencas + ((n - media) ** 2)
desvio = math.sqrt(soma_diferencas / len(lista))
print("Média:", media)
print("Desvio Padrão:", desvio)
# um for pra somar tudo e achar a media dividindo pelo tamanho. dps fiz um outro for subtraindo a media de cada num e elevando ao quadrado (** 2) pra achar o desvio padrao.
```

```python
# Exercício 48
lista = [12, 11, 13, 5, 6]
for i in range(1, len(lista)):
    chave = lista[i]
    j = i - 1
    while j >= 0 and chave < lista[j]:
        lista[j + 1] = lista[j]
        j = j - 1
    lista[j + 1] = chave
print("Ordenado por Inserção:", lista)
# insertion sort feito manual. salvei o valor na variavel chave e fui com o while empurrando as variaveis que sao maiores pro lado direito ate abrir espaço pra ela.
```
