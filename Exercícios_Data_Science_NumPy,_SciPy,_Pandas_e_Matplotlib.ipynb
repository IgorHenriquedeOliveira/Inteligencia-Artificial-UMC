```python
# Exercício 1
import numpy as np
from scipy import integrate
def f(x):
    return (x**2) * np.sin(x)
resultado, erro = integrate.quad(f, 0, np.pi)
print("Resultado da integral:", resultado)
# usei a funçao quad para calcular a integral definida e passei o np.pi como limite final.
```

```python
# Exercício 2
from scipy.integrate import solve_ivp
def eq_diferencial(x, y):
    return x - y
solucao = solve_ivp(eq_diferencial, [0, 10], [1])
print("Solução em x=10:", solucao.y[0][-1])
# o solve_ivp pede a funçao, o intervalo de 0 a 10 e a condição inicial que é o [1].
```

```python
# Exercício 3
from scipy.optimize import root

def f_raiz(x):
    return x**4 - 3*x**3 + 2

raiz = root(f_raiz, 0)
print("Raiz encontrada:", raiz.x)
# a função root acha o ponto onde a conta zera, só precisei chutar um valor inicial (coloquei 0).
```

```python
# Exercício 4
from scipy.optimize import minimize

def f_max(x):
    return -(-x**2 + 4*x + 1)

resultado = minimize(f_max, 0)
print("Ponto de máximo em x =", resultado.x)
# como o scipy tem a funçao minimize, para maximizar é só multiplicar a formula toda por -1.
```

```python
# Exercício 5
import numpy as np
from scipy.fft import fft

t = np.linspace(0, 1, 500)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
frequencias = fft(sinal)
print("Transformada de Fourier calculada com sucesso!")
# usei o fft do scipy igualzinho no exemplo do professor, somente troquei a formula do sinal para juntar o seno e cosseno que ele pediu.
```

```python
# Exercício 6
import numpy as np
from scipy import linalg

matriz5 = np.random.rand(5, 5)
inversa = linalg.inv(matriz5)
print("Matriz Inversa:\n", inversa)
# criei a matriz 5x5 com o numpy e passei ela pro linalg.inv do scipy pra inverter.
```

```python
# Exercício 7
import numpy as np
from scipy.interpolate import interp1d

x_dados = np.array([0, 1, 2, 3, 4])
y_dados = np.array([0, 2, 3, 5, 4])
f_cubica = interp1d(x_dados, y_dados, kind='cubic')

x_novos = np.array([0.5, 1.5, 2.5])
print("Valores interpolados cubicos:", f_cubica(x_novos))
# coloquei o kind='cubic' pra ele fazer a interpolaçao cubica ao inves da linear.
```

```python
# Exercício 8
import numpy as np
from scipy import linalg

A = np.array([[3, 2], [1, 4]])
B = np.array([5, 6])
solucao = linalg.solve(A, B)
print("Solução [x, y]:", solucao)
# separei os coeficientes de um lado (A) e os resultados do outro (B) e mandei o linalg.solve resolver.
```

```python
# Exercício 9
import numpy as np
from scipy.optimize import curve_fit

x_pts = np.array([1, 2, 3, 4, 5])
y_pts = np.array([2, 4, 5, 7, 8])

def parabola(x, a, b, c):
    return a*x**2 + b*x + c

parametros, _ = curve_fit(parabola, x_pts, y_pts)
print("Parâmetros ajustados [a, b, c]:", parametros)
# criei a funçao da parabola e passei pro curve_fit achar os parametros que melhor encaixam nesses pontos.
```

```python
# Exercício 10
from scipy import integrate

def f_dupla(y, x):
    return x*y + x**2

resultado_duplo, erro = integrate.dblquad(f_dupla, 0, 2, lambda x: 0, lambda x: 1)
print("Resultado da integral dupla:", resultado_duplo)
# para scipy fazer integral dupla tem que usar o dblquad, e colocar os limites de y dentro do lambda.
```

```python
# Exercício 11
import numpy as np
from scipy import linalg

matriz4 = np.random.randint(1, 10, (4, 4))
autovalores, autovetores = linalg.eig(matriz4)
print("Autovalores:", autovalores)
# funçao eig() do modulo linalg resolve essa questao de autovalor e autovetor direto.
```

```python
# Exercício 12
import numpy as np
from scipy.optimize import root

def eq_exponencial(x):
    return np.exp(x) - x**2

raiz_exp = root(eq_exponencial, 1)
print("Raiz próxima a 1:", raiz_exp.x)
# mesma coisa do ex 3, usei o root chutando o 1 para achar a raiz da formula.
```

```python
# Exercício 14
from scipy.optimize import minimize

def multivariada(v):
    x, y = v
    return (x - 2)**2 + (y - 3)**2

minimo_multi = minimize(multivariada, [0, 0])
print("Mínimo multivariado [x, y]:", minimo_multi.x)
# como tem duas variaveis (x e y), o scipy pede para passar elas juntas num vetor "v" e depois eu separo dentro da funçao.
```

```python
# Exercício 16
from scipy.integrate import solve_ivp

def eq_segunda_ordem(t, estado):
    y, v = estado
    return [v, -y]

sol_segunda = solve_ivp(eq_segunda_ordem, [0, 10], [1, 0])
print("Solução da segunda ordem:", sol_segunda.y[0][-1])
# o solve_ivp só aceita primeira ordem.  aqui foi dividir a formula em duas variaveis de estado (y e v) para ele conseguir resolver.
```

```python
# Exercício 20
import numpy as np
from scipy import linalg

# Sistema:
# 1x + 1y + 1z = 6
# 2x + 1y - 1z = 1
# 1x - 1y + 1z = 2
A_3 = np.array([[1, 1, 1], [2, 1, -1], [1, -1, 1]])
B_3 = np.array([6, 1, 2])
solucao_3 = linalg.solve(A_3, B_3)
print("Solução [x, y, z]:", solucao_3)
# como o enunciado não deu as formulas prontas, inventei um sistema simples com x, y e z e mandei o linalg.solve achar a resposta.
```

```python
# Exercício 6
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1, 10, 100) # começando de 1 pro log não quebrar com zero
y1 = np.exp(x)
y2 = np.log(x)

plt.subplot(2, 1, 1)
plt.plot(x, y1)
plt.title("Exponencial")

plt.subplot(2, 1, 2)
plt.plot(x, y2)
plt.title("Logaritmo")

plt.tight_layout()
plt.show()
# usei subplot dividindo a tela em duas linhas. precisei começar o linspace do x em 1 porque log de 0 da erro matematico.
```

```python
# Exercício 8
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.fill_between(x, y - 0.2, y + 0.2, alpha=0.3)
plt.show()
# plot faz a linha do seno e o fill_between faz a sombra do intervalo. usei alpha=0.3 para cor ficar transparente e bonita.
```

```python
# Exercício 11
import matplotlib.pyplot as plt
import numpy as np

categorias = ['Cat1', 'Cat2', 'Cat3']
g1 = np.array([5, 10, 15])
g2 = np.array([3, 6, 9])
g3 = np.array([2, 4, 6])

plt.bar(categorias, g1, label='Grupo 1')
plt.bar(categorias, g2, bottom=g1, label='Grupo 2')
plt.bar(categorias, g3, bottom=g1+g2, label='Grupo 3')

plt.legend()
plt.show()
# transformar em array do numpy (np.array) facilita muito a soma das bases (bottom), senão uma barra sobrepoe a outra em vez de empilhar.
```

```python
# Exercício 18
import matplotlib.pyplot as plt

x = ['A', 'B', 'C', 'D']
y = [10, -5, 8, -2]
cores = ['blue' if valor > 0 else 'red' for valor in y]
plt.bar(x, y, color=cores)
plt.show()
# criei uma condição basica para definir a cor com if: se for > 0 vai azul, senao vai vermelho. ai passo pro parametro color.
```
